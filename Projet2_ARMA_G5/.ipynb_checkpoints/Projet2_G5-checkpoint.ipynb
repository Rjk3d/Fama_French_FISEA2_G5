## 1. Présentation du problème et des objectifs

### 1.1 Du statique au dynamique

Dans le Projet 1, nous avons cherché à expliquer le rendement du fonds à l'aide des facteurs de risque du même mois. Cette approche par le modèle de Fama-French était purement statique et descriptive : elle nous indiquait comment le fonds réagissait au marché, mais ne nous permettait pas de faire des prévisions (puisque les facteurs du mois t ne sont connus qu'à la fin de ce même mois).

Pour ce Projet 2, nous changeons d'approche. Nous allons chercher à savoir si les rendements passés contiennent de l'information utile pour prédire les rendements futurs. Pour cela, nous nous appuierons sur les modèles de séries temporelles vus au chapitre 4, et plus particulièrement sur les modèles ARMA.

### 1.2 Les deux questions posées

Notre travail va s'articuler autour de deux grands axes :

Volet A (le lien avec le Projet 1) : Nous allons vérifier si notre modèle de Fama-French a bien capté toute la structure des rendements. S'il reste une dépendance temporelle dans les résidus (un mois qui "annonce" le suivant), cela signifie qu'il manque des informations. À l'inverse, s'ils forment un simple bruit blanc, notre modèle précédent est validé.
Volets B, C et D (le backtesting) : Nous allons calibrer un modèle ARMA sur les 4 premières années de notre échantillon. Notre but est de voir si ce modèle parvient à mieux prévoir les rendements de la dernière année (qui lui sera volontairement cachée pendant l'estimation) qu'une simple prévision naïve basée sur la moyenne historique.

### 1.3 Pour qui cette question a-t-elle un sens ?

En pratique, nos résultats pourraient intéresser plusieurs acteurs :

L'investisseur tactique, qui cherche à anticiper le meilleur moment pour entrer ou sortir du fonds en se basant sur les performances récentes.
Le risk manager, pour qui la fourchette de rendements plausibles (l'intervalle de prévision) est cruciale pour estimer les pertes potentielles du mois suivant.
L'auditeur de notre Projet 1, puisque le volet A nous sert de test de validité (des résidus autocorrélés signifieraient que nos écarts-types calculés par les MCO étaient faussés).

### 1.4 Ce que la théorie financière laisse attendre

D'après l'hypothèse de l'efficience faible des marchés (vue au chapitre 1), toute l'information passée est censée être déjà intégrée dans les prix actuels. Théoriquement, nous ne devrions donc pas pouvoir prévoir les rendements futurs, qui se comporteraient globalement comme un bruit blanc (soit un modèle ARMA(0,0)). Ce projet va donc nous servir de test empirique pour vérifier si cette théorie d'efficience faible s'applique bien à notre fonds.

### 1.5 Hypothèses testées (seuil de 5 %)

| Test | Hypothèse nulle $H_0$ | Ce qu'un rejet signifierait |
|---|---|---|
| Ljung–Box (résidus FF3, ordre 12) | les 12 premières autocorrélations des résidus sont nulles (bruit blanc) | dynamique omise par Fama–French |
| ADF (Dickey–Fuller augmenté) | la série a une racine unitaire (non stationnaire) | la série est stationnaire |
| Ljung–Box (résidus ARMA) | les résidus du modèle ARMA sont un bruit blanc | ordre $(p, q)$ insuffisant |
| Jarque–Bera (résidus ARMA) | résidus normaux | intervalles de prévision gaussiens approximatifs |

## 3. Construction des variables

### 3.1 La série à modéliser : le log-rendement mensuel du fonds

L'énoncé demande de modéliser **la série des rendements du fonds** : on utilise le log-rendement mensuel $r_t$ du Projet 1.
Pourquoi le rendement et pas le prix ? Le chapitre 4 (§4.2.1) l'explique : les **niveaux de prix** (la valeur liquidative)
se comportent comme une **marche aléatoire**, non stationnaire, alors que les **rendements** $r_t = \Delta \ln P_t$ sont
en général **stationnaires**. Un ARMA n'a de sens que sur une série stationnaire. On le vérifiera au §7.2 avec le test ADF.

### 3.2 Le log-prix, pour illustrer la non-stationnarité

On reconstruit le **log de la valeur liquidative** à partir des rendements :
$$\ln\left(\frac{P_t}{P_0}\right) = \sum_{s=1}^{t} r_s ,$$
où $P_0$ est la VL de fin juillet 2021. C'est l'intégrale (la somme cumulée) des rendements ; inversement, le rendement
est la différence première du log-prix. On en déduit aussi un indice de VL **base 100** : $100 \times e^{\sum r_s}$.

### 3.3 Les résidus de Fama–French (volet A)

On reprend les résidus $\hat\varepsilon_t$ exportés par le Projet 1. **Contrôle de cohérence** : on ré-estime le modèle FF3
à partir des colonnes du fichier et l'on vérifie que l'on retrouve exactement les mêmes résidus et les mêmes coefficients.

### 3.4 Découpage entraînement / test, fixé dès maintenant

Pour éviter toute **fuite d'information** (*data leakage*, §4.4.2), le découpage 80 % / 20 % est fixé **avant** tout
traitement : toutes les décisions de calibration (bornes IQR, test ADF, choix de $p$ et $q$, estimation) n'utiliseront que
l'échantillon d'entraînement. Avec 60 mois : **48 mois d'entraînement** (août 2021 – juillet 2025, « les quatre premières
années ») et **12 mois de test** (août 2025 – juillet 2026, « la dernière année »).

```python
# 3.1 Série à modéliser
r = df["r_fonds"].rename("r")

# 3.2 Log-prix et indice de VL base 100 (reconstruits à partir des rendements)
log_vl = r.cumsum().rename("ln(P_t/P_0)")
vl = (100 * np.exp(log_vl)).rename("VL base 100")

# 3.3 Résidus FF3 du Projet 1 + contrôle par ré-estimation
resid_ff3 = df["resid_FF3"]
X3 = sm.add_constant(df[["MKT_RF", "SMB", "HML"]])
ff3 = sm.OLS(df["y"], X3).fit()
print("Coefficients FF3 ré-estimés :", ff3.params.round(4).to_dict())
print(f"Écart maximal entre résidus ré-estimés et résidus exportés : {np.abs(ff3.resid - resid_ff3).max():.2e}")

# 3.4 Découpage chronologique 80 / 20 (jamais de mélange aléatoire sur une série temporelle)
n = len(r)
n_train = int(n * TRAIN_SHARE)
train, test = r.iloc[:n_train], r.iloc[n_train:]
print(f"\nEntraînement : {train.index[0]} -> {train.index[-1]} ({len(train)} mois)")
print(f"Test         : {test.index[0]} -> {test.index[-1]} ({len(test)} mois)")
```

**Lecture.** L'écart entre les résidus ré-estimés et les résidus exportés est de l'ordre de $10^{-16}$, c'est-à-dire
l'erreur d'arrondi de l'ordinateur : le fichier est cohérent avec le Projet 1 (α = −0,0023, β marché = 1,055,
β SMB = −0,098, β HML = −0,245).

Le découpage est **chronologique** : on ne mélange jamais les observations d'une série temporelle, car l'ordre des dates
est précisément l'information que l'on cherche à exploiter.

## 4. Prétraitement

### 4.1 Valeurs manquantes et continuité du calendrier

Pour une série temporelle, il ne suffit pas de compter les `NaN` : il faut aussi vérifier qu'**aucun mois ne manque**
dans l'index. Un mois manquant fausserait silencieusement les retards ($r_{t-1}$ ne serait plus le mois précédent).

### 4.2 Valeurs aberrantes : méthode IQR (§3.6.3), bornes calculées sur l'entraînement

On signale toute observation hors de $[Q_1 - 1{,}5\,IQR\,;\,Q_3 + 1{,}5\,IQR]$. Pour respecter le cloisonnement, les quartiles
sont calculés **sur les 48 mois d'entraînement seulement**, puis appliqués aux deux échantillons. Comme au Projet 1, un point
signalé n'est **pas** supprimé automatiquement : on vérifie d'abord s'il s'agit d'une erreur ou d'un vrai mois de marché.

```python
# 4.1 Valeurs manquantes et mois manquants
print("Valeurs manquantes par colonne :", df.isna().sum().to_dict())
calendrier = pd.period_range(df.index[0], df.index[-1], freq="M")
mois_manquants = calendrier.difference(df.index)
print(f"Mois attendus : {len(calendrier)} | mois présents : {len(df)} | mois manquants : {list(mois_manquants) or 'aucun'}")

# 4.2 Règle IQR, bornes estimées sur l'entraînement uniquement
q1, q3 = train.quantile([0.25, 0.75])
iqr = q3 - q1
borne_basse, borne_haute = q1 - 1.5 * iqr, q3 + 1.5 * iqr
atypiques = r[(r < borne_basse) | (r > borne_haute)]
print(f"\nBornes IQR (entraînement) : [{borne_basse:+.2%} ; {borne_haute:+.2%}]")
print("Mois atypiques :", {str(k): f"{v:+.2%}" for k, v in atypiques.items()} or "aucun")
print(f"Rendement mensuel min / max : {r.min():+.2%} ({r.idxmin()}) / {r.max():+.2%} ({r.idxmax()})")
```

**Bilan du prétraitement.**
* **Aucune valeur manquante et aucun mois manquant** : les 60 mois se suivent sans trou. Aucune imputation, aucune
  suppression.
* **Aucun rendement atypique** au sens de l'IQR, ni dans l'entraînement ni dans le test. Même les mois les plus extrêmes
  (−11,4 % en janvier 2022, +12,1 % en avril 2026) restent dans les moustaches. C'est cohérent avec le Projet 1, qui n'avait
  signalé de points atypiques que sur les facteurs de style, pas sur le fonds. **Toutes les observations sont conservées.**

## 5. Analyse exploratoire

On décrit la série sur l'**entraînement** (ce que le modèle « voit ») et, à titre purement descriptif, sur le **test**. Les
statistiques du test ne servent **à aucune décision** de modélisation : les calculer ici n'est pas une fuite d'information,
les utiliser pour calibrer le modèle en serait une.

Rappels (chapitre 1) : la **skewness** mesure l'asymétrie (0 pour une loi normale), la **kurtosis** l'épaisseur des queues
(3 pour une loi normale). Le test de **Jarque–Bera** combine les deux :
$JB = \frac{n}{6}\left(S^2 + \frac{(K-3)^2}{4}\right) \sim \chi^2(2)$ sous $H_0$ (normalité).

```python
def describe(s):
    jb = stats.jarque_bera(s)
    return pd.Series({"n": len(s), "moyenne mensuelle": s.mean(), "écart-type mensuel": s.std(ddof=1),
                      "moyenne annualisée (x12)": 12 * s.mean(), "volatilité annualisée (x√12)": np.sqrt(12) * s.std(ddof=1),
                      "min": s.min(), "max": s.max(), "skewness": stats.skew(s), "kurtosis": stats.kurtosis(s, fisher=False),
                      "Jarque-Bera": jb.statistic, "p-value JB": jb.pvalue})

desc = pd.DataFrame({"entraînement (48 mois)": describe(train), "test (12 mois)": describe(test),
                     "total (60 mois)": describe(r)})
desc
```

```python
fig, axes = plt.subplots(2, 1, figsize=(11, 6.5), sharex=True)
t_all = r.index.to_timestamp()
t_split = test.index[0].to_timestamp()
for ax in axes:
    ax.axvspan(t_split, t_all[-1] + pd.Timedelta(days=20), color=COLORS["gris"], alpha=0.12)
axes[0].plot(t_all, vl, color=COLORS["fonds"])
axes[0].set(title=f"{TICKER} : valeur liquidative ajustée (base 100 fin juillet 2021) — non stationnaire",
            ylabel="VL base 100")
axes[0].text(t_split + pd.Timedelta(days=15), vl.min(), "test (masqué au modèle)", color=COLORS["gris"], fontsize=9)
axes[1].bar(t_all, r * 100, width=20, color=np.where(r.index < test.index[0], COLORS["fonds"], "black"))
axes[1].axhline(train.mean() * 100, color=COLORS["marche"], ls="--", lw=1.3, label=f"moyenne d'entraînement = {train.mean():.2%}")
axes[1].set(title="Log-rendements mensuels r_t — stationnaires autour d'une moyenne constante", ylabel="% par mois")
axes[1].legend(loc="upper left")
axes[1].xaxis.set_major_formatter(mdates.DateFormatter("%m/%y"))
plt.tight_layout(); save(fig, "fig1_vl_et_rendements"); plt.show()
```

**Lecture de l'analyse exploratoire.**
* **Niveau contre rendement.** Le graphique du haut montre une VL qui **dérive** dans le temps (forte baisse en 2022, puis
  hausse continue) : sa moyenne dépend de la période, elle n'est pas stationnaire. Le graphique du bas montre des
  rendements qui **oscillent autour d'une moyenne stable**, sans tendance : c'est la forme d'une série stationnaire. C'est
  exactement la distinction du chapitre 4 (§4.2.1), que le test ADF va confirmer formellement.
* **Moyenne faible face au bruit.** Sur l'entraînement, le rendement moyen est d'environ **0,87 % par mois** (≈ 10 %/an),
  pour un écart-type d'environ **5,6 % par mois**. Le rapport signal/bruit est donc d'environ 0,16 : d'un mois sur l'autre,
  la variabilité écrase la tendance. Toute prévision mensuelle sera forcément très imprécise.
* **Normalité.** Jarque–Bera ne rejette pas la normalité sur l'entraînement (p ≈ 0,37 ; skewness −0,45, kurtosis 2,6).
  Comme au Projet 1, l'agrégation mensuelle rapproche les rendements d'une loi normale, et 48 points donnent peu de
  puissance au test. Sur les 12 mois de test, la normalité est rejetée (p ≈ 0,008), mais à cause d'**un seul mois**,
  avril 2026 (+12,1 %), qui crée une forte asymétrie positive sur un si petit échantillon.
* **Une dernière année plus faible et plus calme.** Sur le test, le rendement moyen tombe à environ 0,2 % par mois
  (contre 0,87 %) et la volatilité à 15 %/an (contre 19 %/an) : la période de test ne ressemble pas tout à fait à la période
  d'entraînement. Le modèle ne le sait pas, et c'est précisément ce qui rend le backtesting honnête.

## 6. Volet A — Le pont avec le Projet 1 : les résidus de Fama–French sont-ils un bruit blanc ?

### 6.1 Ce que l'on teste

Un **bruit blanc** (§4.2.2) est une série de moyenne nulle, de variance constante, et **sans aucune corrélation entre
dates** : connaître sa valeur passée n'apprend rien sur sa valeur future. Si les résidus du modèle FF3 forment un bruit
blanc, le modèle a capté toute la partie « prévisible à partir du passé » de la relation, et l'hypothèse H3 des MCO
(absence d'autocorrélation des erreurs) est vérifiée.

On mesure la dépendance temporelle par la **fonction d'autocorrélation** (FAC) : $\rho_h = \mathrm{Corr}(\hat\varepsilon_t,
\hat\varepsilon_{t-h})$. Le **test de Ljung–Box** (§4.3.6) teste conjointement la nullité des $m$ premières
autocorrélations :
$$Q_{LB} = n(n+2)\sum_{h=1}^{m}\frac{\hat\rho_h^2}{n-h} \;\sim\; \chi^2(m) \text{ sous } H_0 .$$
Plus les autocorrélations estimées sont grandes, plus $Q_{LB}$ est grand. On rejette $H_0$ (bruit blanc) si la p-value est
inférieure à 5 %. L'énoncé demande $m = 12$ (une année de retards mensuels) ; on affiche aussi les ordres 1 à 11.

**Règle de décision de l'énoncé :** p > 0,05 ⇒ le modèle de Fama–French est validé ; p < 0,05 ⇒ une dynamique
temporelle a été omise, ce qui motive le volet B.

```python
lb_ff3 = acorr_ljungbox(resid_ff3, lags=range(1, LB_LAGS + 1), return_df=True)
lb_ff3["bruit blanc (5%)"] = np.where(lb_ff3["lb_pvalue"] > SEUIL, "oui", "non")
lb12 = lb_ff3.loc[LB_LAGS]
print(f"Ljung-Box à l'ordre {LB_LAGS} sur les résidus FF3 : Q = {lb12.lb_stat:.3f}, p-value = {lb12.lb_pvalue:.4f}")
print("=> Résidus compatibles avec un bruit blanc : modèle Fama-French VALIDÉ" if lb12.lb_pvalue > SEUIL
      else "=> Autocorrélation résiduelle : dynamique omise, passer au volet B")
lb_car = acorr_ljungbox(df["resid_Carhart"], lags=[LB_LAGS], return_df=True).iloc[0]
print(f"(contrôle) Ljung-Box ordre {LB_LAGS} sur les résidus Carhart : Q = {lb_car.lb_stat:.3f}, p-value = {lb_car.lb_pvalue:.4f}")
lb_ff3.T
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
plot_acf(resid_ff3, lags=LB_LAGS, ax=axes[0], color=COLORS["fonds"], vlines_kwargs={"colors": COLORS["fonds"]})
plot_pacf(resid_ff3, lags=LB_LAGS, ax=axes[1], method="ywm", color=COLORS["fonds"], vlines_kwargs={"colors": COLORS["fonds"]})
axes[0].set(title="FAC des résidus FF3", xlabel="retard h (mois)")
axes[1].set(title="FACP des résidus FF3", xlabel="retard h (mois)")
plt.tight_layout(); save(fig, "fig2_volet_A_correlogramme_residus_FF3"); plt.show()
```

### 6.2 Complément : un ARMA trouverait-il quelque chose dans ces résidus ?

Pour confirmer, on estime les 16 modèles ARMA$(p,q)$ avec $p, q \in \{0,1,2,3\}$ **sur les résidus** et on regarde
lequel minimise le BIC. Si les résidus sont un bruit blanc, le meilleur modèle doit être l'ARMA$(0,0)$, c'est-à-dire
« pas de dynamique ». La fonction `grille_arma` définie ici sera réutilisée au volet B.

```python
def grille_arma(serie, max_p=MAX_P, max_q=MAX_Q):
    """Estime ARMA(p, q) pour toute la grille et renvoie AIC, BIC et l'indicateur de convergence de chaque modèle."""
    lignes = []
    for p, q in itertools.product(range(max_p + 1), range(max_q + 1)):
        try:
            fit = ARIMA(serie, order=(p, 0, q)).fit()
            lignes.append({"p": p, "q": q, "k = p+q+1": p + q + 1, "AIC": fit.aic, "BIC": fit.bic,
                           "convergé": bool((fit.mle_retvals or {}).get("converged", True))})
        except Exception:                      # un ordre non estimable est simplement écarté
            lignes.append({"p": p, "q": q, "k = p+q+1": p + q + 1, "AIC": np.nan, "BIC": np.nan, "convergé": False})
    return pd.DataFrame(lignes)

grille_resid = grille_arma(resid_ff3)
print("Ordre minimisant le BIC sur les résidus FF3 :", tuple(grille_resid.loc[grille_resid["BIC"].idxmin(), ["p", "q"]]))
print("Ordre minimisant l'AIC sur les résidus FF3 :", tuple(grille_resid.loc[grille_resid["AIC"].idxmin(), ["p", "q"]]))
grille_resid.sort_values("BIC").head(4).set_index(["p", "q"])
```

**Conclusion du volet A : le modèle de Fama–French du Projet 1 est validé.**
* Ljung–Box à l'ordre 12 : **Q ≈ 10,0, p ≈ 0,61 > 0,05**. On ne rejette pas l'hypothèse de bruit blanc. Aucun ordre
  intermédiaire (1 à 11) ne rejette non plus. Les résidus du modèle Carhart donnent la même conclusion.
* Sur le corrélogramme, aucune autocorrélation ne sort de la bande de confiance à 95 % (zone bleutée), ni pour la FAC ni
  pour la FACP : **aucune signature AR ou MA**.
* La grille ARMA sur les résidus choisit **ARMA(0,0) par le BIC**. L'AIC, moins sévère, préfère de justesse un AR(1)
  (écart d'AIC de 0,3 seulement, donc non significatif), ce qui correspond à la petite autocorrélation d'ordre 1
  (≈ 0,19) déjà vue avec le Durbin–Watson du Projet 1 (1,59).

**Conséquences.** (1) Les écarts-types MCO du Projet 1 sont valides : il n'était pas nécessaire de les corriger par
Newey–West. (2) Le volet A ne révèle **pas de dynamique omise**. Le volet B n'est donc pas « motivé » par un défaut du
Projet 1. Il pose une question différente et plus exigeante : les résidus sont blancs **une fois les facteurs du mois
connus**, mais un investisseur ne connaît pas à l'avance les facteurs du mois prochain. Il ne dispose que du **passé du
fonds**. Le volet B teste donc si ce passé suffit à prévoir le rendement.

## 7. Volet B — Backtesting : découpage, stationnarité et calibration de l'ARMA

### 7.1 Le découpage entraînement / test (rappel du §3.4)

|---|---|---|---|
| Entraînement (80 %) | août 2021 – juillet 2025 | 48 | test ADF, corrélogramme, choix de $(p,q)$ par AIC/BIC, estimation des paramètres |
| Test (20 %) | août 2025 – juillet 2026 | 12 | **uniquement** pour comparer prévisions et réalité au volet D |

Toutes les cellules de cette section n'utilisent que `train`.

### 7.2 Stationnarité : le test de Dickey–Fuller augmenté (ADF)

**Pourquoi ?** Estimer un modèle, c'est supposer que le passé ressemble assez au futur pour qu'on puisse en tirer des
leçons. Si la moyenne ou la variance d'une série dérive dans le temps, ce qu'on apprend sur le passé ne vaut plus pour le
futur. Un processus est **stationnaire au sens faible** si sa moyenne, sa variance et ses autocovariances $\mathrm{Cov}(y_t, y_{t-h})$
ne dépendent pas de la date $t$ (§4.2.1).

**Le test.** On estime la régression augmentée
$$\Delta y_t = c + \delta\, y_{t-1} + \sum_{j=1}^{p} \gamma_j \Delta y_{t-j} + \varepsilon_t , \qquad \delta = \phi_1 - 1 .$$
* $H_0 : \delta = 0$, c'est-à-dire $\phi_1 = 1$ : **racine unitaire**, la série est une marche aléatoire, non stationnaire ;
* $H_1 : \delta < 0$ : la série est **stationnaire** (elle revient vers sa moyenne).

Les retards $\Delta y_{t-j}$ (choisis automatiquement par AIC) « blanchissent » les résidus : c'est le « augmenté ». Le test est
**unilatéral à gauche** : on rejette $H_0$ seulement si la statistique est **suffisamment négative**, c'est-à-dire inférieure
à la valeur critique, ou si la p-value est inférieure à 5 %. La statistique ne suit pas une loi de Student mais la loi de
Dickey–Fuller, d'où des valeurs critiques spécifiques (environ −2,9 à 5 %).

On applique le test :
1. au **log-prix** $\ln P_t$ (avec constante, puis avec constante et tendance, car un prix d'actif a une dérive) : on attend
   une racine unitaire ;
2. au **log-rendement** $r_t$ (avec constante) : on attend la stationnarité.

```python
def adf(serie, regression, nom):
    stat, pval, nlags, nobs, crit, _ = adfuller(serie, autolag="AIC", regression=regression)
    return {"série": nom, "régression": {"c": "constante", "ct": "constante + tendance"}[regression],
            "stat ADF": stat, "p-value": pval, "retards retenus": nlags,
            "VC 1%": crit["1%"], "VC 5%": crit["5%"], "VC 10%": crit["10%"],
            "conclusion (5%)": "stationnaire (H0 rejetée)" if pval < SEUIL else "racine unitaire (H0 non rejetée)"}

log_vl_train = log_vl.loc[train.index]
adf_tab = pd.DataFrame([adf(log_vl_train, "c", "log-prix ln P_t (train)"),
                        adf(log_vl_train, "ct", "log-prix ln P_t (train)"),
                        adf(train, "c", "log-rendement r_t (train)")]).set_index(["série", "régression"])
adf_tab
```

**Lecture du test ADF.**
* **Log-prix : racine unitaire non rejetée** (p ≈ 0,98 avec constante, p ≈ 0,59 avec constante et tendance). La
  statistique ADF est même **positive** avec constante : pas question de rejeter, le test étant unilatéral à gauche. La VL
  se comporte comme une marche aléatoire : la modéliser directement par un ARMA exposerait à une **régression
  fallacieuse** et à des prévisions sans fondement.
* **Log-rendement : stationnarité nettement acceptée.** La statistique (≈ −5,3) est bien inférieure à la valeur critique à
  1 % (≈ −3,6), p-value ≈ 0,00001. On rejette la racine unitaire.

**Conséquence : on modélise les rendements avec $d = 0$**, c'est-à-dire un ARMA$(p,q)$ sur $r_t$. C'est équivalent à un
ARIMA$(p,1,q)$ sur le log-prix (§4.3.3) : le log-prix est **intégré d'ordre 1**, et sa différence première, le rendement,
est stationnaire. Pas besoin de différencier une seconde fois.

### 7.3 Corrélogramme FAC / FACP de l'entraînement : quelle structure chercher ?

Le corrélogramme est l'outil visuel de base pour deviner l'ordre d'un ARMA (§4.2.3). Règle de lecture croisée :

| Signature | FAC (autocorrélation simple) | FACP (autocorrélation partielle) |
|---|---|---|
| **AR(p)** : $y_t = c + \phi_1 y_{t-1} + \dots + \phi_p y_{t-p} + \varepsilon_t$ | décroît progressivement | **s'annule brutalement après le retard p** |
| **MA(q)** : $y_t = \mu + \varepsilon_t + \theta_1\varepsilon_{t-1} + \dots + \theta_q\varepsilon_{t-q}$ | **s'annule brutalement après le retard q** | décroît progressivement |
| **ARMA(p, q)** | décroît progressivement | décroît progressivement |
| **Bruit blanc** | aucune barre significative | aucune barre significative |

Intuition : la FAC au retard $h$ mesure le lien **total** entre $y_t$ et $y_{t-h}$. La FACP mesure le lien **direct**, une fois
retiré l'effet des retards intermédiaires, comme un coefficient de régression multiple « toutes choses égales par
ailleurs ». Pour un AR(1), $y_t$ dépend de $y_{t-2}$ seulement **à travers** $y_{t-1}$ : la FAC décroît ($\rho_h = \phi^h$) mais la
FACP s'annule dès le retard 2.

La zone colorée est la bande de confiance à 95 % sous l'hypothèse de bruit blanc (≈ $\pm 1{,}96/\sqrt{n}$, soit ±0,28 pour
$n = 48$). On limite l'affichage à 12 retards : avec 48 observations, les autocorrélations de rang élevé sont estimées sur trop
peu de paires pour être fiables.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
plot_acf(train, lags=12, ax=axes[0], color=COLORS["fonds"], vlines_kwargs={"colors": COLORS["fonds"]})
plot_pacf(train, lags=12, ax=axes[1], method="ywm", color=COLORS["fonds"], vlines_kwargs={"colors": COLORS["fonds"]})
axes[0].set(title="FAC des rendements (entraînement, 48 mois)", xlabel="retard h (mois)")
axes[1].set(title="FACP des rendements (entraînement, 48 mois)", xlabel="retard h (mois)")
plt.tight_layout(); save(fig, "fig3_volet_B_correlogramme_rendements"); plt.show()

lb_train = acorr_ljungbox(train, lags=[1, 3, 6, 12], return_df=True)
print("Ljung-Box sur les rendements d'entraînement (H0 : bruit blanc) :")
lb_train
```

**Lecture.** Aucune barre ne sort de la bande de confiance, ni sur la FAC ni sur la FACP. La plus grande autocorrélation
est celle d'ordre 1, **légèrement négative** (≈ −0,11), très loin du seuil de ±0,28. Le test de Ljung–Box confirme : aucune
autocorrélation significative jusqu'à l'ordre 12 (p ≈ 0,79).

C'est la **signature d'un bruit blanc**. Le corrélogramme ne suggère aucun ordre AR ou MA. La lecture visuelle laissant peu
de place, on s'en remet aux critères d'information (§7.4), comme le recommande le cours.

### 7.4 Choix de l'ordre (p, q) par les critères d'information AIC et BIC

On estime les 16 modèles ARMA$(p,q)$, $p,q \in \{0,1,2,3\}$, **sur l'entraînement uniquement**, par maximum de
vraisemblance, puis on compare :
$$AIC = -2\ln \hat L + 2k , \qquad BIC = -2\ln \hat L + k \ln n ,$$
où $\hat L$ est la vraisemblance maximisée et $k = p + q + 1$ le nombre de paramètres (constante incluse). Le premier terme
récompense l'ajustement aux données, le second pénalise la complexité. C'est le **compromis biais–variance** du chapitre 3 :
un ARMA(3,3) collera mieux à l'échantillon, mais il risque de « mémoriser » le bruit (**surapprentissage**) et de mal prévoir.

Le BIC pénalise davantage chaque paramètre ($\ln 48 \approx 3{,}9$ contre 2 pour l'AIC) : il choisit des modèles plus
parcimonieux. **Conformément au coding lab (§4.5.4), on retient l'ordre qui minimise le BIC**, et l'on affiche l'AIC pour
comparaison. Remarque : `statsmodels` compte aussi la variance $\sigma^2$ comme paramètre, ce qui ajoute la même constante à
tous les modèles et ne change pas le classement.

```python
grille = grille_arma(train)
best_bic = grille.loc[grille["BIC"].idxmin()]
best_aic = grille.loc[grille["AIC"].idxmin()]
P_OPT, Q_OPT = int(best_bic.p), int(best_bic.q)
print(f"Ordre minimisant le BIC : ARMA({P_OPT},{Q_OPT})  (BIC = {best_bic.BIC:.2f})  <- modèle retenu")
print(f"Ordre minimisant l'AIC  : ARMA({int(best_aic.p)},{int(best_aic.q)})  (AIC = {best_aic.AIC:.2f})")
print(f"Modèles dont l'optimisation n'a pas convergé : {[(int(a), int(b)) for a, b in grille.loc[~grille['convergé'], ['p', 'q']].values]}")

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for ax, crit in zip(axes, ["AIC", "BIC"]):
    mat = grille.pivot(index="p", columns="q", values=crit)
    im = ax.imshow(mat, cmap="Blues_r")
    ax.grid(False)
    ax.set(xticks=range(MAX_Q + 1), yticks=range(MAX_P + 1), xlabel="q (ordre MA)", ylabel="p (ordre AR)",
           title=f"{crit} des ARMA(p, q) — plus bas = meilleur")
    pm, qm = np.unravel_index(np.nanargmin(mat.values), mat.shape)
    vmin, vmax = np.nanmin(mat.values), np.nanmax(mat.values)
    for i in range(MAX_P + 1):
        for j in range(MAX_Q + 1):
            fonce = (mat.iloc[i, j] - vmin) / (vmax - vmin) < 0.4        # case foncée -> texte blanc
            ax.text(j, i, f"{mat.iloc[i, j]:.1f}" + (" ★" if (i, j) == (pm, qm) else ""), ha="center", va="center",
                    fontsize=9, fontweight="bold" if (i, j) == (pm, qm) else "normal",
                    color="white" if fonce else "black")
plt.tight_layout(); save(fig, "fig4_volet_B_grille_AIC_BIC"); plt.show()

grille.sort_values("BIC").head(6).set_index(["p", "q"])
```

**Lecture de la grille.**
* **AIC et BIC choisissent tous deux l'ARMA(0,0)**, c'est-à-dire le modèle $r_t = \mu + \varepsilon_t$ : une constante plus un
  bruit blanc. Aucun terme AR ou MA n'améliore assez la vraisemblance pour payer sa pénalité.
* Les suivants sont l'MA(1) et l'AR(1), à environ **3,2 points de BIC** du meilleur (et 1,3 à 1,4 point d'AIC). Ajouter un
  paramètre fait gagner très peu de vraisemblance, signe qu'il n'y a pas de dynamique à capter.
* Les modèles les plus riches (ARMA(2,2), (2,3), (3,3)…) sont fortement pénalisés et plusieurs **n'ont pas convergé** : avec
  48 points, l'optimiseur ne parvient pas à identifier 5 à 7 paramètres, et ces modèles sont de toute façon écartés.

Cette conclusion rejoint le corrélogramme et l'hypothèse d'efficience faible : **le passé des rendements mensuels de
FMAGX ne contient pas de structure linéaire exploitable.** On retient l'ARMA(0,0), en gardant l'AR(1) et le MA(1) comme
**modèles concurrents** pour vérifier honnêtement, au volet D, qu'un modèle « avec dynamique » ne ferait pas mieux.

### 7.5 Estimation du modèle retenu et diagnostic de ses résidus

Un modèle ARMA est **bien spécifié** quand ses résidus ne contiennent plus aucune structure temporelle, c'est-à-dire quand
ils se comportent comme un bruit blanc (§4.2.2). On vérifie donc sur les résidus du modèle retenu :
* **Ljung–Box à l'ordre 12** ; les degrés de liberté sont corrigés du nombre de paramètres ARMA estimés
  (`model_df = p + q`, ici 0), car estimer des coefficients AR/MA « blanchit » mécaniquement les résidus ;
* **Jarque–Bera** : les intervalles de confiance de `get_forecast` supposent des chocs gaussiens ;
* **à titre de bonus (hors programme, annexe ARCH du polycopié)** : le test **ARCH–LM**, qui applique la logique de
  Breusch–Godfrey aux **résidus au carré**. Il ne teste pas la prévisibilité du **niveau** (comme Ljung–Box), mais celle de
  l'**amplitude** des chocs (regroupement de volatilité).

On estime aussi les deux concurrents AR(1) et MA(1), pour lire la significativité de leur coefficient.

```python
ORDRE = (P_OPT, 0, Q_OPT)
modele = ARIMA(train, order=ORDRE).fit()
print(modele.summary())

concurrents = {"AR(1)": (1, 0, 0), "MA(1)": (0, 0, 1)}
fits = {f"ARMA({P_OPT},{Q_OPT}) retenu (BIC)": modele}
fits.update({nom: ARIMA(train, order=o).fit() for nom, o in concurrents.items()})

lignes = []
for nom, f in fits.items():
    for param in f.params.index:
        lignes.append({"modèle": nom, "paramètre": param, "estimation": f.params[param],
                       "écart-type": f.bse[param], "p-value": f.pvalues[param]})
pd.DataFrame(lignes).set_index(["modèle", "paramètre"])
```

```python
diag = []
for nom, f in fits.items():
    o = f.model.order
    lb = acorr_ljungbox(f.resid, lags=[LB_LAGS], model_df=o[0] + o[2], return_df=True).iloc[0]
    jb = stats.jarque_bera(f.resid)
    arch_stat, arch_p, _, _ = het_arch(f.resid, nlags=6)
    diag.append({"modèle": nom, "Ljung-Box Q(12)": lb.lb_stat, "p LB": lb.lb_pvalue,
                 "JB": jb.statistic, "p JB": jb.pvalue, "ARCH-LM(6) [bonus]": arch_stat, "p ARCH": arch_p,
                 "AIC": f.aic, "BIC": f.bic})
diag = pd.DataFrame(diag).set_index("modèle")
mu_hat, sigma_hat = modele.params["const"], np.sqrt(modele.params["sigma2"])
print(f"Modèle retenu : r_t = {mu_hat:.4f} + e_t,  sigma = {sigma_hat:.4f}  "
      f"(soit {12 * mu_hat:.1%}/an de rendement moyen et {np.sqrt(12) * sigma_hat:.1%}/an de volatilité)")
diag
```

**Lecture de l'estimation.**
* **Modèle retenu :** $\hat r_t = \hat\mu + \varepsilon_t$ avec $\hat\mu$ ≈ **0,87 % par mois** (≈ 10,4 %/an) et
  $\hat\sigma$ ≈ **5,5 % par mois**. La constante n'est **pas significative** (p ≈ 0,31) : même la moyenne est estimée très
  imprécisément sur 4 ans, son intervalle de confiance à 95 % va d'environ −0,8 % à +2,5 % par mois. L'estimateur du maximum de
  vraisemblance de $\mu$ coïncide presque exactement avec la moyenne empirique de l'entraînement : le modèle retenu
  **se confond avec la prévision naïve**.
* **Concurrents :** le coefficient de l'AR(1) ($\hat\phi_1$ ≈ −0,11, p ≈ 0,46) et celui du MA(1) ($\hat\theta_1$ ≈ −0,12,
  p ≈ 0,43) sont **non significatifs**, de signe négatif (léger effet de retour à la moyenne d'un mois sur l'autre) et petits.
  Les deux sont stationnaires et inversibles ($|\phi_1| < 1$, $|\theta_1| < 1$).

**Diagnostic des résidus.**
* **Ljung–Box : bruit blanc non rejeté** pour les trois modèles (p ≈ 0,79 pour l'ARMA(0,0)). Le modèle retenu est
  **bien spécifié** au sens du chapitre 4 : il ne reste pas de structure temporelle dans la moyenne.
* **Jarque–Bera : normalité non rejetée.** Les intervalles de prévision gaussiens de `get_forecast` sont donc acceptables.
* **ARCH–LM (bonus, hors programme) : effets ARCH détectés** (p ≈ 0,005). L'**amplitude** des chocs est en partie prévisible :
  les grands mouvements arrivent par grappes (2022 très agité, 2024 plus calme). Le test d'hétéroscédasticité affiché par
  `summary()` (H ≈ 0,31, p ≈ 0,02) le confirme : la variance des résidus de la fin de l'échantillon vaut environ le tiers
  de celle du début. Le cours l'explique : un modèle peut être
  bien spécifié pour la **moyenne** (Ljung–Box) et mal spécifié pour la **variance** (ARCH). L'ARMA suppose une variance
  constante : ses intervalles de prévision seront donc **trop larges en période calme et trop étroits en période agitée**.
  On garde ce point pour le bilan critique.

## 8. Volet C — Prévision sur l'horizon de test et intervalles de confiance à 95 %

### 8.1 Ce que calcule `get_forecast`

Le modèle estimé sur les 48 mois d'entraînement produit, **en une seule fois depuis juillet 2025**, les prévisions des
12 mois suivants $\hat r_{T+h|T}$, $h = 1, \dots, 12$, sans jamais voir les rendements réalisés du test. Pour chaque horizon, il
donne aussi l'écart-type de l'erreur de prévision, d'où l'intervalle à 95 % :
$$\hat r_{T+h|T} \pm 1{,}96 \times \hat\sigma_{h} .$$

**À quoi s'attendre pour un ARMA(0,0) ?** Comme $r_t = \mu + \varepsilon_t$ et que les chocs futurs sont imprévisibles, la
meilleure prévision est $\hat\mu$ **à tous les horizons** : une droite horizontale. Et l'erreur de prévision est le seul
choc $\varepsilon_{T+h}$, donc $\hat\sigma_h = \hat\sigma$ pour tout $h$ : l'intervalle a **la même largeur à tous les horizons**.
Le « tunnel » qui s'élargit avec $h$ (§4.4.3) n'apparaît que lorsqu'il y a de la dynamique (AR, MA) : les erreurs se cumulent
alors d'un horizon à l'autre. Pour un AR(1), il s'élargit, mais très vite (dès $h = 2$) jusqu'à la variance totale de la série.
On verra au §9.5 que le tunnel réapparaît dès qu'on raisonne sur la **valeur liquidative**, qui cumule les rendements.

```python
H = len(test)
fc = modele.get_forecast(steps=H)
prev = pd.Series(fc.predicted_mean.values, index=test.index, name="prévision")
ic = pd.DataFrame(fc.conf_int(alpha=ALPHA_IC).values, index=test.index, columns=["borne basse 95%", "borne haute 95%"])

tab_prev = pd.concat([test.rename("réalisé"), prev, ic], axis=1)
tab_prev["écart-type de prévision"] = fc.se_mean.values
tab_prev["erreur (réalisé - prévu)"] = tab_prev["réalisé"] - tab_prev["prévision"]
tab_prev["dans l'IC 95%"] = np.where((test >= ic.iloc[:, 0]) & (test <= ic.iloc[:, 1]), "oui", "NON")
couverture = (tab_prev["dans l'IC 95%"] == "oui").mean()
print(f"Modèle ARMA{ORDRE} estimé sur {train.index[0]} -> {train.index[-1]}, prévision de {test.index[0]} à {test.index[-1]}")
print(f"Taux de couverture empirique de l'IC à 95 % : {couverture:.0%} ({int(couverture * H)}/{H} mois)")

# Prévisions des modèles concurrents (même protocole)
prev_conc = {nom: pd.Series(f.get_forecast(steps=H).predicted_mean.values, index=test.index)
             for nom, f in fits.items() if nom in concurrents}
tab_prev
```

**Lecture.**
* La prévision vaut **0,87 % pour chacun des 12 mois**, avec un intervalle à 95 % d'environ **[−10,0 % ; +11,7 %]**,
  identique à tous les horizons. C'est la traduction chiffrée du résultat du volet B : le modèle n'a rien d'autre à
  proposer que la moyenne.
* **Couverture :** 11 mois sur 12 tombent dans l'intervalle (92 %, pour 95 % attendus). Le seul mois hors de l'intervalle est
  **avril 2026 (+12,1 %)**, un mois de très forte hausse du marché (+9,95 % pour Mkt-RF). Avec 12 points, 11/12 est
  parfaitement compatible avec une couverture théorique de 95 %.
* **Largeur de l'intervalle :** environ 22 points de pourcentage pour une prévision de 0,9 point. L'intervalle est
  **25 fois plus large** que la prévision n'est éloignée de zéro : le modèle dit honnêtement qu'il ne sait pas si le mois
  prochain sera positif ou négatif.

## 9. Volet D — Confrontation graphique et métriques d'erreur

### 9.1 Le graphique de confrontation

Un seul graphique superpose, comme le demande l'énoncé : (1) la courbe d'entraînement, (2) les rendements réels de la
période de test, (3) les prévisions du modèle, (4) la zone ombrée de l'intervalle de confiance à 95 %.

```python
fig, ax = plt.subplots(figsize=(12, 5))
t_tr, t_te = train.index.to_timestamp(), test.index.to_timestamp()
ax.plot(t_tr, train * 100, color=COLORS["fonds"], lw=1.6, label="entraînement (48 mois, utilisé pour l'estimation)")
ax.plot(t_te, test * 100, color="black", marker="o", lw=1.6, label="réalisé sur la période de test (masqué au modèle)")
ax.plot(t_te, prev * 100, color=COLORS["marche"], ls="--", marker="s", ms=4,
        label=f"prévision ARMA({P_OPT},{Q_OPT}) (get_forecast, 12 pas)")
ax.fill_between(t_te, ic.iloc[:, 0] * 100, ic.iloc[:, 1] * 100, color=COLORS["marche"], alpha=0.15,
                label="intervalle de confiance à 95 %")
ax.axvline(t_te[0] - pd.Timedelta(days=15), color=COLORS["gris"], ls=":", lw=1.2)
ax.axhline(0, color=COLORS["gris"], lw=0.8)
ax.set(title=f"{TICKER} : backtesting de l'ARMA({P_OPT},{Q_OPT}) — entraînement, réalité et prévision",
       ylabel="log-rendement mensuel (%)")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%m/%y"))
ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.08), ncol=2, fontsize=9)
plt.tight_layout(); save(fig, "fig5_volet_D_confrontation"); plt.show()
```

### 9.2 Métriques d'erreur et comparaison à la prévision naïve

Sur les $H = 12$ mois de test (§4.4.3) :
$$RMSE = \sqrt{\frac{1}{H}\sum_{h=1}^{H}(r_{T+h} - \hat r_{T+h|T})^2}, \qquad MAE = \frac{1}{H}\sum_{h=1}^{H}|r_{T+h} - \hat r_{T+h|T}| .$$
Le **RMSE** élève les erreurs au carré : il pénalise fortement les grosses erreurs. Le **MAE** est l'erreur moyenne en valeur
absolue, directement lisible en points de rendement. Un RMSE n'a **aucun sens seul** : il faut le comparer à des prévisions
de référence. On en utilise trois :
* **naïve « moyenne historique »** (référence de l'énoncé) : la moyenne des 48 mois d'entraînement, pour tous les mois ;
* **naïve « dernière valeur »** : le rendement de juillet 2025 répété (la persistance) ;
* **rendement nul** : ne rien prévoir du tout.

Un ratio RMSE modèle / RMSE naïf **inférieur à 1** signifierait une valeur ajoutée.

```python
def metriques(reel, prevu):
    e = reel - prevu
    return {"RMSE": np.sqrt(np.mean(e ** 2)), "MAE": np.mean(np.abs(e))}

previsions = {
    f"ARMA({P_OPT},{Q_OPT}) retenu (BIC)": prev,
    **{nom: s for nom, s in prev_conc.items()},
    "Naïf : moyenne historique (train)": pd.Series(train.mean(), index=test.index),
    "Naïf : dernière valeur (juillet 2025)": pd.Series(train.iloc[-1], index=test.index),
    "Rendement nul": pd.Series(0.0, index=test.index),
}
metr = pd.DataFrame({nom: metriques(test, s) for nom, s in previsions.items()}).T
ref = metr.loc["Naïf : moyenne historique (train)"]
metr["RMSE / RMSE naïf"] = metr["RMSE"] / ref["RMSE"]
metr["MAE / MAE naïf"] = metr["MAE"] / ref["MAE"]
print(f"Écart entre la prévision ARMA et la moyenne d'entraînement : {prev.iloc[0] - train.mean():+.6f} (en décimal)")

err2 = (test - prev) ** 2
print(f"Part de la somme des erreurs au carré due au seul mois d'avril 2026 : {err2.loc['2026-04'] / err2.sum():.0%}")
metr
```

**Lecture des métriques.**
* **ARMA(0,0) et prévision naïve sont indiscernables** : RMSE ≈ **4,23 %** et MAE ≈ **2,91 %** par mois pour les deux (ratio
  = 1,000). C'est logique : la prévision de l'ARMA(0,0) est l'estimation de la moyenne par maximum de vraisemblance,
  quasiment identique à la moyenne empirique (écart de l'ordre de 0,001 point). **Le modèle n'apporte aucune valeur
  ajoutée prédictive.**
* **Les concurrents AR(1) et MA(1) ne font pas mieux** : leurs prévisions ne s'écartent de la moyenne que le premier
  mois (pour le MA(1), l'effet disparaît exactement après un mois ; pour l'AR(1), il est divisé par environ 10 chaque mois),
  et leurs métriques sont pratiquement identiques à celles de la moyenne (ratios de 0,998 et 0,999).
* **La persistance (« dernière valeur ») est la pire** (RMSE ≈ 4,7 %) : répéter le rendement de juillet 2025 suppose une
  inertie que les données démentent (FAC d'ordre 1 légèrement négative).
* **Le « rendement nul » fait un peu mieux que la moyenne** (RMSE ≈ 4,19 %). Ce n'est pas une prouesse : la dernière année a été
  plus faible que les quatre précédentes (0,2 % par mois contre 0,87 %), donc prévoir 0 tombe, par chance, plus près. Rien
  dans l'entraînement ne permettait de l'anticiper.
* **Le RMSE est dominé par un seul mois** : avril 2026 représente à lui seul plus de la moitié de la somme des erreurs au
  carré. Sur 12 points, une seule surprise de marché suffit à faire bouger les classements. Les écarts entre modèles
  (quelques centièmes de point) sont **sans signification statistique**.

**Conclusion du volet D :** sur la dernière année, aucun modèle ARMA ne bat la moyenne historique. Le meilleur prédicteur
du rendement mensuel de FMAGX, à partir de son seul passé, est sa moyenne de long terme, et cette prévision reste
extrêmement imprécise (erreur typique ≈ 4 points par mois).

### 9.3 Pour aller plus loin : backtest glissant à un pas (ré-estimation mensuelle)

Le protocole de l'énoncé prévoit les 12 mois **en une seule fois** depuis juillet 2025. En pratique, un investisseur
**met à jour** son modèle chaque mois, avec le rendement qu'il vient d'observer. On simule donc une **fenêtre croissante** :
pour chaque mois de test $t$, on ré-estime le modèle (avec l'ordre choisi sur l'entraînement) sur **toutes les données
jusqu'à $t-1$ seulement**, et on prévoit $t$. Il n'y a **aucune fuite d'information** : chaque prévision n'utilise que le passé
disponible à sa date.

Références glissantes : la moyenne de tous les mois passés (moyenne « croissante ») et le dernier rendement observé
(marche aléatoire sur les prix). On ajoute le **taux de bon sens directionnel** (part des mois où le signe prévu est le
bon), plus parlant pour un investisseur. La boucle sur les dates est ici indispensable (une ré-estimation par mois).

```python
def backtest_glissant(serie, n_train, ordre):
    """Prévision à 1 pas avec ré-estimation sur une fenêtre croissante : la prévision du mois k n'utilise que serie[:k]."""
    prevs = [ARIMA(serie.iloc[:k], order=ordre).fit().forecast(1).iloc[0] for k in range(n_train, len(serie))]
    return pd.Series(prevs, index=serie.index[n_train:])

glissant = {f"ARMA({P_OPT},{Q_OPT}) glissant": backtest_glissant(r, n_train, ORDRE),
            "AR(1) glissant": backtest_glissant(r, n_train, (1, 0, 0)),
            "MA(1) glissant": backtest_glissant(r, n_train, (0, 0, 1)),
            "Naïf : moyenne croissante": r.expanding().mean().shift(1).loc[test.index],
            "Naïf : dernier rendement": r.shift(1).loc[test.index]}

metr_gl = pd.DataFrame({nom: {**metriques(test, s), "bon signe": np.mean(np.sign(s) == np.sign(test))}
                        for nom, s in glissant.items()}).T
metr_gl["RMSE / RMSE naïf"] = metr_gl["RMSE"] / metr_gl.loc["Naïf : moyenne croissante", "RMSE"]
print(f"Part de mois positifs sur le test : {(test > 0).mean():.0%}")
metr_gl
```

```python
fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(t_te, test * 100, color="black", marker="o", label="réalisé")
for (nom, s), c in zip(list(glissant.items())[:3], [COLORS["marche"], COLORS["autre"], COLORS["fonds"]]):
    ax.plot(t_te, s * 100, marker=".", lw=1.4, color=c, label=nom)
ax.plot(t_te, glissant["Naïf : dernier rendement"] * 100, color=COLORS["gris"], ls=":", label="naïf : dernier rendement")
ax.axhline(0, color=COLORS["gris"], lw=0.8)
ax.set(title="Backtest glissant à 1 pas (ré-estimation chaque mois sur le seul passé)", ylabel="% par mois")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%m/%y")); ax.legend(fontsize=8, ncol=2)
plt.tight_layout(); save(fig, "fig6_backtest_glissant"); plt.show()
```

**Lecture.** Le backtest glissant confirme le volet D, dans des conditions plus réalistes :
* les prévisions de l'ARMA(0,0) glissant **sont** la moyenne croissante (RMSE identiques) ;
* l'AR(1) et le MA(1) glissants restent collés à la moyenne : leurs prévisions oscillent de quelques dixièmes de point
  autour d'elle, sans améliorer le RMSE de façon notable ; leur MAE est même un peu moins bon ;
* la marche aléatoire (« le mois prochain ressemblera au dernier ») est nettement la pire (RMSE ≈ 6,4 %) ;
* **le signe est mal prévu** : les modèles « moyenne » prévoient toujours une hausse, alors que seuls 5 mois sur 12 ont été
  positifs (42 % de bons signes). Les modèles avec dynamique font encore moins bien (33 %). Un investisseur qui aurait
  utilisé ces prévisions pour entrer ou sortir du fonds n'aurait rien gagné.

### 9.4 Pourquoi on ne choisit jamais le modèle sur l'échantillon de test (*data snooping*)

On calcule, **à titre d'illustration seulement**, le RMSE de test des 16 modèles de la grille. Si l'on choisissait le modèle
qui a le meilleur RMSE de test, on utiliserait l'information du test pour calibrer le modèle : c'est exactement la **fuite
d'information** que le cours interdit (§4.4.2), et la performance affichée serait trompeusement optimiste.

```python
rmse_grille = []
for p, q in itertools.product(range(MAX_P + 1), range(MAX_Q + 1)):
    f = ARIMA(train, order=(p, 0, q)).fit()
    rmse_grille.append({"p": p, "q": q, "RMSE test (12 pas)": metriques(test, f.get_forecast(H).predicted_mean.values)["RMSE"]})
snoop = grille.merge(pd.DataFrame(rmse_grille), on=["p", "q"])
snoop["rang BIC"] = snoop["BIC"].rank().astype(int)
snoop["rang RMSE test"] = snoop["RMSE test (12 pas)"].rank().astype(int)
print(f"Corrélation de rang (Spearman) entre BIC d'entraînement et RMSE de test : "
      f"{stats.spearmanr(snoop['BIC'], snoop['RMSE test (12 pas)']).statistic:+.2f}")
snoop.sort_values("RMSE test (12 pas)").head(5).set_index(["p", "q"])[["BIC", "convergé", "RMSE test (12 pas)", "rang BIC", "rang RMSE test"]]
```

**Lecture.** Le « meilleur » modèle sur le test serait un **ARMA(2,3)**, avec un RMSE de 3,89 % contre 4,23 % pour le modèle
retenu. Mais c'est l'un des **plus mauvais modèles selon le BIC** (rang 12 sur 16), et son optimisation **n'a même pas
convergé**. Les trois « meilleurs » modèles sur le test sont d'ailleurs tous des modèles surparamétrés dont l'estimation n'a
pas convergé. Leurs prévisions oscillent au hasard, et il se trouve que ces oscillations ont collé, par chance, à quelques
mois de la dernière année. Rien ne dit qu'elles feraient de même sur l'année suivante.

Le classement par BIC sur l'entraînement et le classement par RMSE sur le test sont presque sans lien (corrélation de
rang ≈ +0,12). Avec 12 points de test, le RMSE est trop bruité pour départager des modèles aussi proches. **La seule démarche honnête est celle que nous avons
suivie** : choisir l'ordre sur l'entraînement (BIC), puis mesurer une seule fois la performance sur le test.

### 9.5 Traduction en valeur liquidative : la prévision face à l'évolution réelle du fonds et du marché

Pour un investisseur, la question concrète est : « combien vaudront 100 $ placés dans le fonds fin juillet 2025 ? ». Comme le
log-prix est la somme des rendements, la prévision du log-prix à l'horizon $h$ est la **somme des prévisions de rendement**,
et son erreur est la **somme des erreurs** :
$$\ln\frac{P_{T+h}}{P_T} - \sum_{j=1}^{h}\hat r_{T+j|T} = \sum_{j=1}^{h} e_{T+j} .$$
Pour un ARMA, l'erreur à l'horizon $j$ s'écrit $e_{T+j} = \sum_{i=0}^{j-1}\psi_i\,\varepsilon_{T+j-i}$, où les $\psi_i$ sont les
coefficients de la représentation MA$(\infty)$ du modèle (fonction `arma2ma`). En regroupant les termes, la variance de l'erreur
cumulée vaut $\sigma^2\sum_{k=0}^{h-1}\Psi_k^2$, avec $\Psi_k = \psi_0 + \dots + \psi_k$. Pour l'ARMA(0,0), $\psi_0 = 1$ et les autres sont nuls :
la variance vaut $h\,\sigma^2$, et l'écart-type croît comme $\sqrt{h}$. **Le tunnel de prévision du cours réapparaît** : sur la VL,
l'incertitude s'accumule avec l'horizon.

On superpose le marché américain (Mkt-RF + RF de K. French) investi au même moment, pour situer le fonds « face au marché ».

```python
def prevision_cumulee(fit, h, alpha=ALPHA_IC):
    """Prévision du log-rendement cumulé sur 1..h mois et son IC, via les poids psi de la forme MA(infini)."""
    moy = np.cumsum(fit.get_forecast(h).predicted_mean.values)
    psi = arma2ma(fit.polynomial_ar, fit.polynomial_ma, lags=h)       # psi_0, ..., psi_{h-1}
    var = fit.params["sigma2"] * np.cumsum(np.cumsum(psi) ** 2)        # sigma^2 * somme des Psi_k^2
    z = stats.norm.ppf(1 - alpha / 2)
    return moy, moy - z * np.sqrt(var), moy + z * np.sqrt(var)

cum_moy, cum_bas, cum_haut = prevision_cumulee(modele, H)
base = 100
vl_reelle = base * np.exp(test.cumsum())
vl_prev, vl_bas, vl_haut = (base * np.exp(x) for x in (cum_moy, cum_bas, cum_haut))
marche = base * (1 + df.loc[test.index, "MKT_RF"] + df.loc[test.index, "RF"]).cumprod()

t0 = train.index[-1].to_timestamp()
tt = [t0] + list(t_te)
fig, ax = plt.subplots(figsize=(11, 4.5))
ax.fill_between(tt, [base, *vl_bas], [base, *vl_haut], color=COLORS["marche"], alpha=0.15, label="IC à 95 % de la VL prévue")
ax.plot(tt, [base, *vl_prev], color=COLORS["marche"], ls="--", label=f"VL prévue par l'ARMA({P_OPT},{Q_OPT})")
ax.plot(tt, [base, *vl_reelle], color="black", marker="o", ms=4, label=f"VL réelle de {TICKER}")
ax.plot(tt, [base, *marche], color=COLORS["gris"], marker=".", label="marché US (K. French), pour comparaison")
ax.set(title="100 $ investis fin juillet 2025 : prévision vs réalité sur l'année de test", ylabel="valeur (base 100)")
ax.xaxis.set_major_formatter(mdates.DateFormatter("%m/%y")); ax.legend(loc="upper left", fontsize=9)
plt.tight_layout(); save(fig, "fig7_tunnel_VL"); plt.show()

print(f"Au bout de 12 mois : VL prévue = {vl_prev[-1]:.1f}  [IC 95 % : {vl_bas[-1]:.1f} ; {vl_haut[-1]:.1f}]")
print(f"                     VL réelle = {vl_reelle.iloc[-1]:.1f} | marché US = {marche.iloc[-1]:.1f}")
```

**Lecture.**
* Le modèle prévoyait qu'au bout d'un an, 100 $ deviendraient environ **111 $**, avec un intervalle à 95 % très large,
  d'environ **76 $ à 162 $** : c'est le tunnel du cours, qui s'ouvre comme $\sqrt{h}$.
* En réalité, le fonds a fini l'année vers **103 $** (+2,6 %), bien en dessous de la prévision centrale mais **largement dans
  l'intervalle**. Le point bas de mars 2026 (≈ 90 $) reste lui aussi dans le tunnel.
* **Face au marché**, le constat est sévère : le marché américain a fini l'année vers **119 $** (+19 %), soit environ
  **16 points de mieux** que le fonds. Cette sous-performance rejoint l'alpha hors échantillon très négatif mesuré au
  Projet 1 (§8 : −11,5 %/an sur la même année, une fois le style du fonds pris en compte). Un modèle univarié du fonds ne
  pouvait pas l'anticiper : rien dans les 48 mois d'entraînement ne l'annonçait.
* Ce graphique résume le projet : la **prévision centrale** n'apporte rien de plus que « le fonds rapporte en moyenne
  ≈ 10 % par an ». C'est l'**intervalle** qui est informatif pour l'investisseur : sur un an, l'issue plausible va d'une perte
  d'environ un quart du capital à un gain d'environ 60 %.

```python
# ============ SYNTHÈSE CHIFFRÉE ============
print(f"{FUND_NAME} ({TICKER}) — entraînement {train.index[0]}->{train.index[-1]} (n={len(train)}), "
      f"test {test.index[0]}->{test.index[-1]} (n={len(test)})")
print(f"Volet A : Ljung-Box(12) résidus FF3 : Q = {lb12.lb_stat:.2f}, p = {lb12.lb_pvalue:.3f} -> modèle FF3 validé")
print(f"Volet B : ADF log-prix p = {adf_tab['p-value'].iloc[0]:.3f} (non stationnaire) | ADF rendements p = {adf_tab['p-value'].iloc[2]:.5f} (stationnaire)")
print(f"          Ordre BIC = ARMA({P_OPT},{Q_OPT}), ordre AIC = ARMA({int(best_aic.p)},{int(best_aic.q)}) | "
      f"Ljung-Box résidus p = {diag['p LB'].iloc[0]:.3f} | ARCH-LM p = {diag['p ARCH'].iloc[0]:.3f}")
print(f"Volet C : prévision = {prev.iloc[0]:.2%}/mois, IC 95 % = [{ic.iloc[0, 0]:.1%} ; {ic.iloc[0, 1]:.1%}], couverture = {couverture:.0%}")
r_m, r_n = metr.iloc[0], metr.loc["Naïf : moyenne historique (train)"]
print(f"Volet D : RMSE ARMA = {r_m.RMSE:.4f} vs naïf = {r_n.RMSE:.4f} | MAE ARMA = {r_m.MAE:.4f} vs naïf = {r_n.MAE:.4f}")

# Export des résultats (tableaux réutilisables dans le rapport)
tab_prev.to_csv(RES_DIR / "previsions_12_mois.csv")
metr.to_csv(RES_DIR / "metriques_volet_D.csv")
metr_gl.to_csv(RES_DIR / "metriques_backtest_glissant.csv")
grille.to_csv(RES_DIR / "grille_AIC_BIC.csv", index=False)
print("\nFigures :", sorted(p.name for p in FIG_DIR.glob("*.png")))
print("Résultats :", sorted(p.name for p in RES_DIR.glob("*.csv")))
```

## 10. Bilan critique

### 10.1 Réponse aux questions posées

1. **Volet A — le modèle de Fama–French du Projet 1 est validé.** Ses résidus sont compatibles avec un bruit blanc
   (Ljung–Box à l'ordre 12 : p ≈ 0,61 ; aucune autocorrélation significative ; ARMA(0,0) choisi par le BIC). Le modèle
   statique ne laisse pas de dynamique exploitable, et ses écarts-types MCO sont valides.
2. **Volet B — il n'y a pas de structure ARMA dans les rendements du fonds.** Le log-prix est non stationnaire (ADF :
   p ≈ 0,98), le rendement est stationnaire (p < 0,001), et sur l'entraînement **AIC et BIC choisissent tous deux
   l'ARMA(0,0)** : le rendement mensuel se résume à une moyenne (≈ 0,87 %/mois) plus un bruit blanc (≈ 5,5 %/mois).
3. **Volets C et D — aucune valeur ajoutée prédictive.** La prévision ARMA est une droite à 0,87 %/mois, avec un
   intervalle à 95 % d'environ ±11 points. Sur la dernière année, son RMSE (4,23 %) et son MAE (2,91 %) sont **identiques**
   à ceux de la moyenne historique. Les concurrents AR(1) et MA(1) ne font pas mieux, ni en prévision à 12 pas ni en
   backtest glissant. L'intervalle de confiance, lui, est **bien calibré** (11 mois sur 12 dedans).

**Interprétation économique.** Ce résultat « négatif » est le résultat attendu et il est **informatif** : il est cohérent
avec l'**hypothèse d'efficience faible** des marchés (chapitre 1). Si les rendements mensuels d'un grand fonds actions
diversifié pouvaient être prévus par leur propre passé, les investisseurs exploiteraient cette régularité jusqu'à la
faire disparaître. Concrètement :
* **pour un investisseur**, il n'y a pas de « bon moment » statistique pour entrer dans FMAGX à partir de ses performances
  récentes : la meilleure estimation du mois prochain est la moyenne de long terme, avec une incertitude énorme ;
* **pour un risk manager**, la partie utile du modèle est l'**intervalle** (la dispersion), pas la prévision centrale ;
* **les deux projets se répondent** : le rendement de FMAGX est expliqué à 94 % par les facteurs **du même mois**
  (Projet 1), mais il est imprévisible à partir de **son passé** (Projet 2). Ce qui fait bouger le fonds, ce sont les chocs
  de marché contemporains, et ces chocs sont, par nature, imprévisibles.

### 10.2 Ce que le modèle ne capture pas : limites assumées

* **Très peu de données.** 48 mois d'entraînement : les tests ADF et Ljung–Box ont une puissance faible, et une
  autocorrélation réelle de ±0,15 passerait inaperçue (seuil de détection ≈ ±0,28). « Pas de dynamique détectée » ne veut
  pas dire « pas de dynamique du tout ».
* **Une seule fenêtre de test de 12 mois.** Les métriques reposent sur 12 erreurs, dont une (avril 2026) pèse plus de la
  moitié du RMSE. Les écarts de RMSE entre modèles ne sont pas statistiquement significatifs, et l'année de test était
  moins favorable que l'entraînement, ce qui avantage par hasard la prévision « rendement nul ».
* **Variance supposée constante.** Le test ARCH–LM (bonus) détecte un regroupement de volatilité (p ≈ 0,005). L'ARMA ignore
  cette dynamique de la variance : ses intervalles sont trop larges en période calme et trop étroits en période agitée.
  La bonne couverture observée (11/12) tient en partie au fait que la dernière année a été plutôt calme.
* **L'incertitude sur les paramètres est ignorée** par `get_forecast` : l'intervalle ne tient compte que des chocs futurs,
  pas de l'imprécision de $\hat\mu$ (dont l'IC à 95 % va de −0,8 % à +2,5 % par mois).
* **Un modèle univarié et linéaire.** L'ARMA n'utilise que le passé du fonds. Il ignore toute information extérieure
  (taux, valorisations, volatilité implicite) et toute non-linéarité (changement de régime).
* **Une fréquence mensuelle.** Les éventuelles dépendances de très court terme (quelques jours) disparaissent par
  agrégation. Elles existent parfois en données quotidiennes, mais seraient de toute façon trop faibles pour couvrir les
  coûts de transaction.
* **Log-rendements et rendements simples.** Comme au Projet 1, la moyenne des log-rendements est inférieure d'environ
  $\sigma^2/2$ (≈ 0,15 %/mois) à celle des rendements simples. Cela ne change pas le classement des modèles, mais il faut le
  garder en tête pour traduire la prévision en performance.

### 10.3 Pistes d'amélioration concrètes

1. **Allonger l'historique et multiplier les fenêtres de test.** FMAGX existe depuis 1963 : 20 ans de données mensuelles
   (240 mois) permettraient un backtest glissant sur plus de 100 mois de test, et de **tester formellement** l'égalité des
   RMSE (test de Diebold–Mariano) au lieu de comparer deux chiffres bruités.
